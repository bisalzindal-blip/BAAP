```python
import os
import sys
import shutil
import subprocess
from pathlib import Path

subprocess.check_call([sys.executable, "-m", "pip", "install", "-q", "biopython", "pandas", "numpy", "requests", "tqdm", "openpyxl"])

baap_path = Path("/content/BAAP")
if baap_path.exists():
    shutil.rmtree(baap_path)

pipeline_path = Path("/content/protease_pipeline")
if pipeline_path.exists():
    shutil.rmtree(pipeline_path)

subprocess.run(["git", "clone", "-q", "https://github.com/bisalzindal-blip/BAAP.git", "/content/BAAP"], check=True)

BAAP_SRC = "/content/BAAP/BAAP/src"
if BAAP_SRC not in sys.path:
    sys.path.insert(0, BAAP_SRC)

BASE_DIR = Path("/content/protease_pipeline")
for directory in [
    BASE_DIR / "input",
    BASE_DIR / "prokka",
    BASE_DIR / "results",
    BASE_DIR / "logs",
    BASE_DIR / "merops",
    BASE_DIR / "interproscan",
]:
    directory.mkdir(parents=True, exist_ok=True)

print("BAAP directories initialized successfully.")

for module_name in list(sys.modules):
    if module_name == "baap" or module_name.startswith("baap."):
        del sys.modules[module_name]

from baap import run_pipeline
from baap.config import LOG_FILE, DIRS

LOG_FILE.parent.mkdir(parents=True, exist_ok=True)
for path in DIRS.values():
    Path(path).mkdir(parents=True, exist_ok=True)

import baap.merops as merops
print("Loaded merops.py:", merops.__file__)

run_pipeline()
```
