# BAAP - Bacterial Annotation & Analysis of Proteases
## v1.0

BAAP is a bioinformatics pipeline for identifying and classifying proteases from bacterial genomes.

### Workflow:
1. Genome upload
2. Prokka annotation
3. MEROPS DIAMOND screening
4. InterProScan domain validation
5. Evidence-based classification
